```python
from advectionSolver import advectionSolver
from ionqStatePrepare import ionqSP
import numpy as np
from os import getenv
import ResGen
import utils
IONQ_API_KEY    = getenv("IONQ_API_KEY")

import matplotlib.pyplot as plt
dim = 2
size = 10
tList = np.linspace(0,0.2,3)
print(tList)
sp = ionqSP(size,dim)
sp.GaussDistribution2D()
```

```python
ad = advectionSolver(dim,[1,1])
idList = []
for t in tList:
    print("time: %f, trotter number: %d" % (t,min(t//0.05,1)))
    iq = ad.solve(size,t,"ionq",trotterNum=min(int(t//0.05),1),trotterOrder=2,circ=sp.circuit,onSimulator=True,noise=False)
    idList.append(iq.task['id'])
with open('simulation.txt','w') as f:
    for id in idList:
        f.write(f"{id}\n")
```

```python
idList = []
with open('simulation.txt','r') as f:
    for id in f:
        idList.append(id[:-1])
print(idList)
resList = [ResGen.resfromId(id,dim*size) for id in idList]
resDict = ResGen.resultGen(resList,size,dim)
mat = np.zeros((size,size,len(tList)))
keys = list(resDict.keys())
print(keys)
for tstep in range(len(tList)):
    for inx in range(len(resDict.keys())):
        row = inx // size
        col = inx % size
        mat[row][col][tstep] = resDict[keys[inx]][tstep]

for row in range(size):
    for col in range(size):
        print('%.5f' % mat[row][col][0],end=' ')
    print()

for i in range(len(tList)):
    fig = plt.figure()

    plt.imshow(mat[:,:,i])
    plt.title("Time = %.2f" % tList[i])
    plt.savefig('../Figs/simulation%d.png' % i)
    plt.show()
```

```python
adNoise = advectionSolver(dim,[1,1])
idNoiseList = []
for t in tList:
    print("time: %f, trotter number: %d" % (t,min(t//0.05,1)))
    iq = adNoise.solve(size,t,"ionq",trotterNum=min(int(t//0.05),1),trotterOrder=2,circ=sp.circuit,onSimulator=True,noise=True)
    idNoiseList.append(iq.task['id'])
with open('noise.txt','w') as f:
    for id in idNoiseList:
        f.write(f"{id}\n")
```
