```python
import numpy as np
from scipy.linalg import expm,kron
import matplotlib.pyplot as plt
from matplotlib import cm
plt.rcParams.update({
    "text.usetex": True,
    "font.family": "sans-serif",
    "font.sans-serif": "Helvetica",
})

def recover(phi,N,pList):
    recoverMat = np.zeros((N,N))
    Psize = len(pList)
    phi = phi.reshape((N,N,Psize))
    for x in range(N):
        for y in range(N):
            maxInx = np.argmax(phi[x,y,:])
            recoverMat[x,y] = pList[maxInx]
            #phiNorm = np.linalg.norm(phi[:,x,y])
            #if phiNorm > 1:
            #    phiNow = phi[:,x,y]/phiNorm
            #    phiNormalized = phiNow * phiNow
            #else:
            #    phiNormalized = np.zeros_like(phi[:,x,y])
            #recoverMat[x,y] = np.dot(pList,phiNormalized)
    return recoverMat

def deltaFunc(x,y,a):
    return np.exp(-(np.abs(x-y)/a)**2)/(a*np.sqrt(np.pi))
    #return a/((x-y)**2 + a **2)/np.pi
    #return np.sin(abs(x-y)/a)/np.pi/abs(x-y)

def initfunc(location):
    return 0.4 * (np.sin(2 * np.pi * location[0]) + np.sin(2 * np.pi * location[1]))
    #return 0.5 * (location[0] <= 0.5) + 0.5 * (location[0]>0.5 and location[1] <= 0.5) - 0.5 * (location[0] > 0.5) * (location[1] > 0.5)
    #return np.exp(-0.5*((location[0]-0.5)**2 + (location[1]-0.5)**2)/0.1/0.1)/5
```

```python
# settings
N = 128
Psize = 128
x = [_/N for _ in range(N)]
y = x
X,Y = np.meshgrid(x,y)
del x,y
pList = np.linspace(-1,1,Psize)
uinit = np.zeros((N,N))
for x in range(N):
    for y in range(N):
        uinit[x,y] = initfunc([x/N,y/N])
phiinit = np.zeros((N,N,Psize))
for pInx in range(len(pList)):
    p = pList[pInx]
    for x in range(N):
        for y in range(N):
            phiinit[x,y][pInx] = deltaFunc(p,uinit[x,y],3/N)
phiinit = phiinit.reshape(Psize*N**2,1)

recoverMat = recover(phiinit,N,pList)
fig, ax = plt.subplots(subplot_kw={"projection": "3d"})
surf = ax.plot_surface(X, Y, uinit, cmap=cm.coolwarm,
                linewidth=0, antialiased=False)
plt.show()

fig, ax = plt.subplots(subplot_kw={"projection": "3d"})
surf = ax.plot_surface(X, Y, recoverMat, cmap=cm.coolwarm,
                linewidth=0, antialiased=False)
plt.show()

print(np.linalg.norm(recoverMat-uinit)/np.linalg.norm(uinit))
```
